```python
def college_chatbot_with_score(question):

    cleaned_question = clean_text(question)

    question_vector = vectorizer.transform(
        [cleaned_question]
    )

    similarities = cosine_similarity(
        question_vector,
        knowledge_vectors
    )[0]

    best_index = similarities.argmax()
    best_score = similarities[best_index]

    if best_score < 0.15:
        return (
            "Sorry, I couldn't find relevant information "
            "in my college knowledge base.",
            best_score
        )

    answer = college_df.iloc[best_index]["answer"]

    return answer, best_score


print("✅ Confidence score enabled!")
```

```python
def smart_college_chatbot(question):

    cleaned_question = clean_text(question)

    question_vector = vectorizer.transform(
        [cleaned_question]
    )

    similarities = cosine_similarity(
        question_vector,
        knowledge_vectors
    )[0]

    best_index = similarities.argmax()
    best_score = similarities[best_index]

    if best_score < 0.20:
        return (
            "Sorry, I don't have enough information about that "
            "in my college knowledge base."
        )

    return college_df.iloc[best_index]["answer"]


print("✅ Smart chatbot created!")
```

```python
print("🎓 COLLEGE STUDENT ASSISTANT")
print("=" * 60)
print("Ask me questions about the college.")
print("Type 'exit' to close the chatbot.")
print("=" * 60)

while True:

    question = input("\n👤 You: ").strip()

    if question.lower() == "exit":
        print("\n🤖 Assistant: Thank you! Have a great day.")
        break

    if question == "":
        print("🤖 Assistant: Please enter a question.")
        continue

    answer = smart_college_chatbot(question)

    print("🤖 Assistant:", answer)
```
